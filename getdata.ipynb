```python
import h5py

file_path = "/Users/baixuezhang/Documents/SJSU/study/CS271/Project/jurkat.h5"

with h5py.File(file_path, "r") as f:
    print("Top-level keys:")
    for k in f.keys():
        print("-", k)

    def show_group(name, obj):
        if isinstance(obj, h5py.Dataset):
            print(f"DATASET: {name}, shape={obj.shape}, dtype={obj.dtype}")
        else:
            print(f"GROUP:   {name}")

    print("\nFull structure:")
    f.visititems(show_group)
```

```python
import h5py

file_path = "/Users/baixuezhang/Documents/SJSU/study/CS271/Project/jurkat.h5"

with h5py.File(file_path, "r") as f:
    for key in ["gene", "target_gene"]:
        print(f"\n=== obs/{key} ===")
        obj = f["obs"][key]

        if isinstance(obj, h5py.Group):
            print("This is a group.")
            print("Subkeys:", list(obj.keys()))
            for subk in obj.keys():
                print(f"  - {subk}: shape={obj[subk].shape}, dtype={obj[subk].dtype}")

                if subk == "categories":
                    vals = obj[subk][:]
                    print("  First 20 categories:")
                    for v in vals[:20]:
                        print("   ", v)

                if subk == "codes":
                    vals = obj[subk][:]
                    print("  First 20 codes:")
                    print(vals[:20])

        else:
            print("This is a dataset.")
            print("shape:", obj.shape, "dtype:", obj.dtype)
            vals = obj[:]
            print("First 20 values:")
            print(vals[:20])
```
